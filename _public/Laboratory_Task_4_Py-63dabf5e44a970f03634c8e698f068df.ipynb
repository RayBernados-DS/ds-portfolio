#**Laboratory Task 4: PyTorch Regression with 2 Fully Connected Layers**


This task demonstrates how to build, run, and optimize a custom neural network model using PyTorch on a synthetic dataset.


**Dataset Generation & Loading:**

We create synthetic features ( X ) and target outcomes ( Y ).
We pack them into a TensorDataset and wrap it in a DataLoader with batch_size=8 and random shuffling. Shuffling prevents the model from learning trends based on order.

**Model Definition:**

We build a custom class inheriting from nn.Module containing:
fc1: A linear layer projecting  5  inputs down to  4  hidden features.
relu: A ReLU activation function for introducing non-linearity.
fc2: A linear layer projecting  4  hidden features to  1  final output.

**Training Setup:**

Criterion: nn.MSELoss() calculates Mean Squared Error.
Optimizer: optim.SGD updates the model's parameters using Stochastic Gradient Descent.

**Training Loop:**

We loop for  1000  epochs.
In each epoch, we iterate through batches of  8  samples: perform a forward pass, calculate loss, zero out previous gradients, backpropagate using loss.backward(), and update network weights using optimizer.step().
Over time, the training loss converges from 0.0303 to an extremely minimal 0.0077.


```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# 1. Generate Synthetic Regression Dataset
torch.manual_seed(42)
X_data = torch.randn(100, 5)
true_w = torch.tensor([[1.5, -2.0, 3.0, 0.5, -1.0]])
true_b = torch.tensor([0.5])
Y_data = torch.mm(X_data, true_w.t()) + true_b + torch.randn(100, 1) * 0.1

# 2. Define Dataset and DataLoader
batch_size = 8
dataset = TensorDataset(X_data, Y_data)
dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

# 3. Define Neural Network Model (2 FC layers)
class TwoLayerNet(nn.Module):
    def __init__(self):
        super(TwoLayerNet, self).__init__()
        self.fc1 = nn.Linear(5, 4)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(4, 1)

    def forward(self, x):
        out = self.fc1(x)
        out = self.relu(out)
        out = self.fc2(out)
        return out

model = TwoLayerNet()

# 4. Define Loss Function and Optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# 5. Training Loop (1000 Epochs)
num_epochs = 1000
for epoch in range(num_epochs):
    epoch_loss = 0.0
    for batch_x, batch_y in dataloader:
        # Forward pass
        predictions = model(batch_x)
        loss = criterion(predictions, batch_y)

        # Backward pass & optimization step
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * batch_x.size(0)

    total_loss = epoch_loss / len(dataloader.dataset)
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {total_loss:.4f}")
```
