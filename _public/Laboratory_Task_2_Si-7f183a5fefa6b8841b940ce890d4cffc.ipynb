```python
import numpy as np

# Inputs, Targets, and Parameters
x = np.array([1.0, 0.0, 1.0])
y = np.array([1.0])

W1 = np.array([
    [0.2, -0.3],
    [0.4, 0.1],
    [-0.5, 0.2]
])  # shape: (3, 2)

W2 = np.array([-0.3, -0.2])  # shape: (2,)

theta_1 = -0.4
theta_2 = 0.2
theta_3 = 0.1

# Activation function: ReLU
def relu(z):
    return np.maximum(0, z)

# Hidden Layer computation
# Z_1 = (w11 * x1) + (w13 * x2) + (w15 * x3) + theta_1
Z1 = (W1[0, 0] * x[0]) + (W1[1, 0] * x[1]) + (W1[2, 0] * x[2]) + theta_1
# Z_2 = (w12 * x1) + (w14 * x2) + (w16 * x3) + theta_2
Z2 = (W1[0, 1] * x[0]) + (W1[1, 1] * x[1]) + (W1[2, 1] * x[2]) + theta_2

H1 = relu(Z1)
H2 = relu(Z2)

# Output Layer computation
Z3 = (W2[0] * H1) + (W2[1] * H2) + theta_3
y_hat = relu(Z3)

# Squared Error calculation
error = 0.5 * (y - y_hat) ** 2

print(f"Z1: {Z1:.4f}, H1 (ReLU): {H1:.4f}")
print(f"Z2: {Z2:.4f}, H2 (ReLU): {H2:.4f}")
print(f"Z3 (Output): {Z3:.4f}, y_hat (ReLU): {y_hat:.4f}")
print(f"Error (Loss): {error[0]:.6f}")

```

#**Single Forward Pass & Error Calculation**


This task manually computes how data flows through a basic feedforward neural network from the input layer, through one hidden layer, to the final output prediction, utilizing the **ReLU** activation function.

1. **Define Inputs ($x$), Target ($y$), and Parameters:**
   - Inputs: $x = [1.0, 0.0, 1.0]$
   - Target: $y = [1.0]$
   - Biases: $\theta_1 = -0.4$ (for hidden neuron 1), $\theta_2 = 0.2$ (for hidden neuron 2), and $\theta_3 = 0.1$ (for output neuron).

2. **Hidden Layer Summation ($Z_n$) and Activation ($H_n$):**
   - We compute the weighted sum of inputs plus bias for both hidden units:
     $$Z_1 = (w_{11} \cdot x_1) + (w_{13} \cdot x_2) + (w_{15} \cdot x_3) + \theta_1$$
     $$Z_1 = (0.2 \cdot 1.0) + (0.4 \cdot 0.0) + (-0.5 \cdot 1.0) - 0.4 = -0.7$$
   - Next, apply the **ReLU (Rectified Linear Unit)** activation function, which outputs $0$ if the input is negative, and the input value itself if positive ($f(z) = \max(0, z)$):
     $$H_1 = \max(0, -0.7) = 0.0$$
   
   - Repeat for the second hidden neuron:
     $$Z_2 = (w_{12} \cdot x_1) + (w_{14} \cdot x_2) + (w_{16} \cdot x_3) + \theta_2$$
     $$Z_2 = (-0.3 \cdot 1.0) + (0.1 \cdot 0.0) + (0.2 \cdot 1.0) + 0.2 = 0.1$$
     $$H_2 = \max(0, 0.1) = 0.1$$

3. **Output Layer Summation ($Z_3$) and Final Prediction ($\hat{y}$):**
   - Calculate the weighted sum of the hidden layer activations plus the output bias:
     $$Z_3 = (w_{21} \cdot H_1) + (w_{22} \cdot H_2) + \theta_3$$
     $$Z_3 = (-0.3 \cdot 0.0) + (-0.2 \cdot 0.1) + 0.1 = 0.08$$
   - Apply ReLU activation to find the network's prediction $\hat{y}$:
     $$\hat{y} = \max(0, 0.08) = 0.08$$

4. **Calculate Squared Error:**
   - We evaluate how close our prediction was using the squared error formula:
     $$\text{Error} = \frac{1}{2}(y - \hat{y})^2 = \frac{1}{2}(1.0 - 0.08)^2 = 0.4232$$
