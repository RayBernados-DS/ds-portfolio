```python
# 1. Perform Standard Imports
import torch
import numpy as np


# 2. Create a function called `set_seed()` that accepts `seed: int` as a parameter
def set_seed(seed: int):
    np.random.seed(seed)
    torch.manual_seed(seed)


# 3. Create a NumPy array called "arr" that contains 6 random integers between 0 (inclusive) and 5 (exclusive),
# call the `set_seed()` function and use `42` as the seed parameter.
set_seed(42)
arr = np.random.randint(0, 5, size=6)
print("3. arr:\n", arr)

# 4. Create a tensor "x" from the array above
x = torch.from_numpy(arr)
print("\n4. x:\n", x)

# 5. Change the dtype of x from int32 to int64
x = x.type(torch.int64)
print("\n5. x dtype:\n", x.dtype)

# 6. Reshape x into a 3x2 tensor
x = x.reshape(3, 2)
print("\n6. Reshaped x (3x2):\n", x)

# 7. Return the right-hand column of tensor x
right_column = x[:, 1]
print("\n7. Right-hand column:\n", right_column)

# 8. Without changing x, return a tensor of square values of x
x_squared = x.pow(2)  # or x ** 2 / torch.square(x)
print("\n8. Square values of x:\n", x_squared)

# 9. Create a tensor y with the same number of elements as x (6 elements),
# that can be matrix-multiplied with x. Shape must be 2x3 to multiply with (3x2).
set_seed(42)
y = torch.randint(0, 5, (2, 3))
print("\n9. y (2x3):\n", y)

# 10. Find the matrix product of x and y
matrix_product = torch.mm(x, y)  # or x @ y
print("\n10. Matrix product (x @ y):\n", matrix_product)
```
